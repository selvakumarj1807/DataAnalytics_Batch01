# DataFrame
-----------

DataFrame is most commonly used object in pandas. It is a table like datastructure containing row and columns similar to excel spredsheet.

```python
# Define the Data

weather_data = {
    'day': ['1/1/2017', '2/1/2017', '3/1/2017', '4/1/2017', '1/1/2017'],
    'temperature':[32, 35, 28, 32, 31],
    'windspeed': [6, 7, 2, 7, 4],
    'event': ['Rain', 'Sunny', 'Snow', 'Rain', 'Sunny']
}

print(weather_data, type(weather_data))
```

```python
# Create DataFrame

import pandas as pd

df = pd.DataFrame(weather_data)

df
```

# Create Csv File Using DataFrame Values

```python
df.to_csv('weather_data.csv', index=False)

print('File Created Successfully!...')
```

# Read CSV File

```python
readFile = pd.read_csv('weather_data.csv')

readFile
```
