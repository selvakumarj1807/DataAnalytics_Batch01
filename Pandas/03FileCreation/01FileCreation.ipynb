# Create CSV Files

```python
import pandas as pd
```

```python
# Define the Data

weather_data = {
    'day': ['1/1/2017', '2/1/2017', '3/1/2017', '4/1/2017', '1/1/2017'],
    'temperature':[32, 35, 28, 32, 31],
    'windspeed': [6, 7, 2, 7, 4],
    'event': ['Rain', 'Sunny', 'Snow', 'Rain', 'Sunny']
}

print(weather_data, type(weather_data))
```

```python
# Create DataFrame

df = pd.DataFrame(weather_data)

df
```

```python
# DataFrame Values to CSV Values

fileName = 'weather_data.csv'

df.to_csv(fileName, index=False)

print(fileName, ' File Created Successfully!...')
```

# Create Microsoft Excel File

```python
# Create Python Variables

df_stocks = pd.DataFrame({
    'tickers': ['GOOGL', 'WMT', 'MSFT'],
    'price': [845, 65, 64], 
    'pe': [30.37, 14.26, 30.97],
    'eps': [27.82, 4.61, 2.12]
})

weather_data = pd.DataFrame({
    'day': ['1/1/2017', '2/1/2017', '3/1/2017', '4/1/2017', '1/1/2017'],
    'temperature':[32, 35, 28, 32, 31],
    'windspeed': [6, 7, 2, 7, 4],
    'event': ['Rain', 'Sunny', 'Snow', 'Rain', 'Sunny']
})


print(df_stocks)

print('----------------------------------------------------------------')

print(weather_data)
```

# Create Excel File

If have a No Module error(openpyxl). Then run Command in CMD

pip install openpyxl

```python
with pd.ExcelWriter('stocks_wheather.xlsx') as writer:
    df_stocks.to_excel(writer, sheet_name="stocks")
    weather_data.to_excel(writer, sheet_name="weather")

print('File Created Successfully!...')
```
