# Different ways of Creating DataFrame

# 1 Using Dictionary

```python
import pandas as pd
```

```python
# Define the Data

weather_data = {
    'day': ['1/1/2017', '2/1/2017', '3/1/2017', '4/1/2017', '1/1/2017'],
    'temperature':[32, 35, 28, 32, 31],
    'windspeed': [6, 7, 2, 7, 4],
    'event': ['Rain', 'Sunny', 'Snow', 'Rain', 'Sunny']
}

print(weather_data, type(weather_data))
```

```python
# Create DataFrame

df01 = pd.DataFrame(weather_data)

df01
```

# 2. Using Tuples & List

```python
# Define The Data

weather_data = [
    ('1/1/2017', 32, 6, 'Rain'),
    ('2/1/2027', 31, 9, 'Sunny'),
    ('1/3/2017', 30, 2, 'Snow')
]

print(weather_data, type(weather_data))
```

```python
# Create DataFrame

df02 = pd.DataFrame(data=weather_data, columns=['day', 'temperature', 'windspeed', 'event'])

df02
```

# Using List of Dictionary

```python
# Define The Data

weather_data = [
    {'day': '1/1/2017', 'temperature': 32,'windspeed': 6, 'event': 'Rain'},
    {'day':'2/1/2027', 'temperature': 31,'windspeed': 9, 'event': 'Sunny'},
    {'day':'1/3/2017', 'temperature': 30,'windspeed': 2, 'event': 'Snow'}
]

print(weather_data, type(weather_data))
```

```python
# Create DataFrame

df03 = pd.DataFrame(data=weather_data, columns=['day', 'temperature', 'windspeed', 'event'])

df03
```
