# 3. Python Connection Code

```python
import mysql.connector

# Connect to MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="company"
)

print("Connected successfully!")

cursor = conn.cursor()
```

# 4. Create Table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(50),
    salary INT
)
""")

print("Table created")
```

# 5. Insert Data

```python
sql = "INSERT INTO employees (name, salary) VALUES (%s, %s)"
values = ("John", 50000)

cursor.execute(sql, values)
conn.commit()

print("Data inserted")
```

# 6. Fetch Data

```python
cursor.execute("SELECT * FROM employees")

for row in cursor.fetchall():
    print(row, type(row))

    print(row[0], ' : ', row[1], ' has salary is ', row[2])

    print('------------')
```

# 7. Update Data

```python
cursor.execute("UPDATE employees SET salary = 60000 WHERE name = 'John'")
conn.commit()

print('Updated Successfully!...')
```

# 8. Delete Data

```python
cursor.execute("DELETE FROM employees WHERE name = 'John'")
conn.commit()

print('Delete Successfully!...')
```

# Full Working Code

```python
import mysql.connector

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="company"
)

cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(50),
    salary INT
)
""")

# Insert
cursor.execute("INSERT INTO employees (name, salary) VALUES (%s, %s)", ("Selva", 70000))
conn.commit()

# Fetch
cursor.execute("SELECT * FROM employees")
print(cursor.fetchall())

cursor.close()
conn.close()
```
