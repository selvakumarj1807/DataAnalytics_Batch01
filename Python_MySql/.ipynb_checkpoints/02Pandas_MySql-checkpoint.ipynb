# Using mysql-connector-python + Pandas

```python
import mysql.connector
import pandas as pd

# Connect to MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="company"
)

# Query
query = "SELECT * FROM employees"

# Load into DataFrame
df = pd.read_sql(query, conn)

# Show data
print(df)

# Close connection
conn.close()
```

# Using SQLAlchemy

More scalable & widely used in real projects

# Install

pip install sqlalchemy pymysql

```python
import pandas as pd
from sqlalchemy import create_engine

# Create connection engine
# create_engine("mysql+pymysql://user_name:your_password@localhost/your_database")
engine = create_engine("mysql+pymysql://root:@localhost/company")

# Query
query = "SELECT * FROM employees"

# Load into DataFrame
df = pd.read_sql(query, engine)

df
```

```python
df = pd.read_sql("SELECT name FROM employees", engine)

df
```
