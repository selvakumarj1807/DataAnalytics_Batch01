# Types of Plots

# 01. Line Plots

A line plot shows data points connected by straight lines.

It is mainly used to show trends or changes over time.

```python
x = [1, 2, 3]

y = [4, 5, 6]

plt.plot(x, y, linewidth=1, color='green', marker='*', linestyle='--') # 'r*--', linewidth=1
plt.title('Simple Line Chart')
plt.xlabel('X - Axis')
plt.ylabel('Y - Axis')
plt.grid()
plt.show()
```

```python
x = [1, 2, 3]

y = [4, 5, 6]

plt.plot(x, y, linewidth=1, color='green', marker='o', linestyle='--', label='data') # 'r*--', linewidth=1
plt.title('Simple Line Chart')
plt.xlabel('X - Axis')
plt.ylabel('Y - Axis')
plt.grid()
plt.legend(loc='lower right')
plt.show()
```

```python
x = [1, 2, 3]

y = [4, 5, 6]

z = [7, 8, 9]

plt.plot(x, y, linewidth=1, color='green', marker='o', linestyle='--', label='x & y -data') 
plt.plot(x, z, linewidth=1, color='red', marker='o', linestyle='--', label='x & z data') 
plt.plot(y, z, linewidth=1, color='blue', marker='o', linestyle='--', label='y & zdata') 

plt.title('Simple Line Chart')
plt.xlabel('X - Axis')
plt.ylabel('Y - Axis')
plt.grid()
plt.legend(loc='lower right')
plt.show()
```

# 02 Bar charts

A bar chart represents data using rectangular bars, where:

1. Height(or length) of each bar = value
2. Each bar = One Cateory

Used to Compare values between categories

```python
x = [1, 2, 3]

y = [4, 5, 6]

plt.bar(x, y)
plt.title('Simple Bar Chart')
plt.xlabel('X - Axis')
plt.ylabel('Y - Axis')
plt.show()
```

```python
x = ['A', 'B', 'C', 'D']

y = [10, 20, 30, 25]

plt.bar(x, y, color='orange', edgecolor='red', label='Data')
plt.legend()

plt.title('Simple Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

plt.grid(axis='y')

for i in range(len(x)):
    plt.text(x[i], y[i], y[i])

plt.show()
```

# 03 Scatter plots

A Scatter plot is used to show the relationship between two variables by plotting points on a graph.

```python
x = ['A', 'B', 'C', 'D']

y = [10, 20, 30, 25]

plt.scatter(x, y, color='red', alpha=0.5, s=100, edgecolor='blue')

plt.grid()

for i in range(len(x)):
    plt.text(x[i], y[i] + 1.5, y[i], ha='center')

plt.show()
```

# 04. Pie charts

A Pie chart is used to show a whole is divided into parts(percentages). Each slice represents a category's contribution.

```python
x = ['A', 'B', 'C', 'D']

y = [10, 20, 30, 25]

plt.pie(y, labels=x) # plt.pie(values, labels=labels)

plt.show()
```

```python
values = [50, 30, 20]

labels = ['Food', 'Rent', 'Savings']

plt.pie(
    values, 
    labels=labels, 
    autopct='%1.1f%%',
    colors=['gold', 'lightblue', 'lightgreen'],
    explode=[0, 0.1, 0.2], 
    shadow=True,
    startangle=90
)

plt.title('Montly Expenses')

plt.show()
```
