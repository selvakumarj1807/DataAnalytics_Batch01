# Numpy

1. Numpy stands for Numerical Python
2. Numpy is a Python Library used for working with arrays.
3. It also has Function for working in domain of linear algebra, Fourier transform, matrices

# Why Use Numpy

1. In python we have lists that serve the purpose of arays, but they are slow to process.
2. Numpy aims to provide an array object that is up to 50x faster than traditional Python List.
3. The array object in Numpy is called ndarray.

# Why is Numpy Faster than Lists?

Numpy arrays are at one continuous place in memory Unlike lists, So processes can access and manipulate then very efficiently.

```python
import numpy as np
```

```python
v1 = [10, 20, 30]

print(v1, type(v1))

print('--------------------------------------')

# Single Dimention array

nv1 = np.array(v1)

print(nv1, type(nv1))
```

```python
v2 = [[10, 20, 30], [40, 50, 60]]

print(v2, type(v2))

print('--------------------------------------')

# Two Dimention array

nv2 = np.array([1, 2 ,3])

print(nv2, type(nv2))
```

# Arange

```python
av1 = np.arange(0, 10)

print(av1)
```

```python
av1 = np.arange(0, 10, 2)

print(av1)
```

# Linspace

```python
lv1 = np.linspace(0, 10, 100)

print(lv1)
```

```python
lv1 = np.linspace(0, 10, 5, dtype=int)

print(lv1)
```
