# Copy Data

```python
import numpy as np
```

```python
v1 = np.linspace(1, 50, 8, dtype=int)

print(v1)
```

```python
v1Copy = v1.copy()

print(v1Copy)
```

```python
print(v1[2])

print(v1[2:6])

print('----------------------------------')

v1[2] = 20

v1[3:6] = 25

print('----------------------------------')

print(v1)
```

```python
print(v1Copy)
```

# Using Arithmatic Operations on Arrays

```python
v1 = np.array([10, 20, 30, 40])

print(v1)

print('------------------------------------')

print(v1 + 2)
```

```python
v2 = np.array([[10, 20], [30, 40]])

print(v2)

print('------------------------------------------------')

print(v2 + 2)
```

```python
v1 = np.array([[1, 2], [3, 4]])


aV1 = v1 + v2

print(aV1)
```

# min, max, argmin, argmax

```python
v1 = np.random.randint(1, 20, 10)

print(v1)
```

```python
print(v1.min())

print(v1.max())
```

```python
print(v1.argmin())

print(v1.argmax())
```

```python
print(v1.sum())
```

```python
print(np.median(v1))
```
