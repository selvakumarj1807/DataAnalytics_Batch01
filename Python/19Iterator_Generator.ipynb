# Python Iterator
-----------------

A Python Iterator is an object that allows you to traverse (loop through) a collection of elements one by one.

# 1. What is an Iterator?

An iterator is an object that implements two methods:

1. __iter__() → returns the iterator object itself

2. __next__() → returns the next value (and raises StopIteration when done)

```python
users = ['user01', 'user02', 'user03', 'user04']

for i in users:
    print(i)
```

# Using Iterator

```python
users = ['user01', 'user02', 'user03', 'user04']

IteratorVariable = iter(users)

print('-----------------------------------')

print(next(IteratorVariable))

print('-----------------------------------')

print(next(IteratorVariable))

print('-----------------------------------')

print(next(IteratorVariable))

print('-----------------------------------')

print(next(IteratorVariable))
```

```python
class CountUp:
    def __init__(self, max):
        self.max = max
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.max: # 3 <= 5
            val = self.current # 3
            self.current += 1 # 4
            return val # 3
        else:
            raise StopIteration

counter = CountUp(5)

for num in counter:
    print(num)

print('---------------------------------')

counter = CountUp(10)

for num in counter:
    print(num)
    print('##')
```

# Python Generator
------------------

A Python Generator is a special type of function that allows you to generate values one at a time,
instead of returning everything at once like a normal function.

```python
def count_up(n):
    i = 1
    while i <= n:
        yield i   # generates value
        i += 1

gen = count_up(5)

print(next(gen))
```

```python
def infinite():
    i = 1
    while True:
        yield i
        i += 1

genInfinate = infinite()

print(next(genInfinate))

print(next(genInfinate))

print(next(genInfinate))
```

```python
def count_up(max):
    current = 1
    while current <= max:
        yield current
        current += 1
        
a = count_up(2)

print(next(a))

print(next(a))

print('----------------------------------------')

for num in count_up(5):
    print(num)
    print('##')
```

```python
gen = (x*x for x in range(5))

for i in gen:
    print(i)
```
